```python
import hashlib, os, datetime, time
print(f'727823TUCY040 | Imports OK | {datetime.datetime.now()}')
```

```python
def compute_hash(filepath, algorithm='sha256'):
    h = hashlib.new(algorithm)
    try:
        with open(filepath, 'rb') as f:
            while chunk := f.read(8192):
                h.update(chunk)
        return h.hexdigest()
    except FileNotFoundError:
        return None

def create_file(path, content):
    with open(path, 'w') as f:
        f.write(content)

print('727823TUCY040 | Helper functions defined')
```

## Test Case 1 — Baseline & Integrity Check (No Tampering)

```python
print(f'727823TUCY040 | TC1 | {datetime.datetime.now()}')
create_file('nb_tc1.txt', 'user=admin\nrole=superuser\nactive=true\n')
baseline_hash = compute_hash('nb_tc1.txt')
current_hash  = compute_hash('nb_tc1.txt')
print(f'Baseline SHA256 : {baseline_hash}')
print(f'Current  SHA256 : {current_hash}')
print(f'Intact          : {baseline_hash == current_hash}')
os.remove('nb_tc1.txt')
```

## Test Case 2 — Detect Content Modification

```python
print(f'727823TUCY040 | TC2 | {datetime.datetime.now()}')
create_file('nb_tc2.py', "print('legitimate script')\n")
baseline_hash = compute_hash('nb_tc2.py')
time.sleep(1)
create_file('nb_tc2.py', "import os; os.system('whoami')  # TAMPERED\n")
current_hash = compute_hash('nb_tc2.py')
print(f'Baseline : {baseline_hash[:32]}...')
print(f'Current  : {current_hash[:32]}...')
print(f'TAMPERED : {baseline_hash != current_hash}')
if baseline_hash != current_hash:
    print('  ⚠ ALERT: HASH_CHANGED — File has been modified!')
os.remove('nb_tc2.py')
```

## Test Case 3 — Detect File Deletion

```python
print(f'727823TUCY040 | TC3 | {datetime.datetime.now()}')
create_file('nb_tc3.cfg', 'db_password=secret\napi_key=abc123\n')
baseline_hash = compute_hash('nb_tc3.cfg')
os.remove('nb_tc3.cfg')
current_hash = compute_hash('nb_tc3.cfg')
print(f'Baseline hash   : {baseline_hash[:32]}...')
print(f'Current hash    : {current_hash}')
if current_hash is None:
    print('  ⚠ ALERT: FILE_DELETED — File no longer exists on disk!')
```
